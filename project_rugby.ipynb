```python
import pandas as pd
from IPython.display import display
from rugbypy.player import fetch_all_players, fetch_player_stats

# 1. Recherche de Romain Ntamack
players = fetch_all_players()
col_name = 'player_name' if 'player_name' in players.columns else 'name'
ntamack_info = players[players[col_name].str.contains("Ntamack", case=False, na=False)]

if not ntamack_info.empty:
    ntamack_id = str(ntamack_info['player_id'].iloc[0])
    df_ntamack = fetch_player_stats(player_id=ntamack_id)

    print("=== 1. APERÇU SYNTHÉTIQUE DE SES MATCHS ===")
    # Sélection des métriques clés pour une vue d'ensemble propre
    cols_principales = [
        col for col in ['player_id', 'name', 'team','game_date', 'team_id', 'position', 'carries', 'line_breaks', 'tackles_completed', 'turnovers_lost', 'turnovers_won', 'kicks', 'passes', 'metres_carried', 'offloads', 'defenders_beaten', 'try_assists', 'tries', 'tackles_made', 'tackles_missed', 'dominant_tackles', 'ruck_turnovers', 'lineouts_won', 'carries_per_minute', 'total_tackles_per_minute', 'red_cards', 'yellow_cards', 'penalties_conceded', 'game_date', 'competition_id', 'competition', 'match_id', 'team_vs', 'team_vs_id', 'weight_kg', 'height_cm', 'clean_breaks', 'conversion_goals', 'drop_goals_converted', 'kick_percent_success', 'kicks_from_hand', 'lineout_won_steal', 'mauls_won', 'meters_run', 'missed_tackles', 'offload', 'penalty_goals', 'points', 'rucks_won', 'runs', 'tackles', 'total_free_kicks_conceded', 'total_lineouts', 'turnover_knock_on', 'turnovers_conceded', 'player_name', 'want_team', 'want_teamvs', 'minutes_played']
 
        if col in df_ntamack.columns
    ]
    display(df_ntamack[cols_principales].head(25))

    print("\n=== 2. FICHE DÉTAILLÉE D'UN MATCH (Toutes les statistiques) ===")
    # On transpose la 1re ligne (.T) pour afficher toutes les variables à la verticale sans défiler à l'horizontale
    match_detail = df_ntamack.head(1).T
    match_detail.columns = ['Valeur']
    display(match_detail)
else:
    print("Joueur non trouvé dans la base.")
```

```python
import pandas as pd
import numpy as np

# 1. Filtrer les matchs du Top 14
top14_teams = [
    "Stade Toulousain", "Clermont Auvergne", "Racing 92", "La Rochelle",
    "Bordeaux Bègles", "Toulon", "Stade Français", "Castres", 
    "LOU Rugby", "Montpellier", "Aviron Bayonnais", "Pau", "USA Perpignan"
]

df_top14 = matches[
    matches['home_team'].isin(top14_teams) | matches['away_team'].isin(top14_teams)
].copy()

# Conversion propre des dates
df_top14['date'] = pd.to_datetime(df_top14['date'], format='%Y%m%d')
df_top14 = df_top14.sort_values('date').reset_index(drop=True)

print("Matchs Top 14 extraits :", len(df_top14))
df_top14.head(14)
```
